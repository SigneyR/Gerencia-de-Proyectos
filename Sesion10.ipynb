1. Diagrama de red lógica

Es la representación gráfica de todas las actividades de un proyecto y sus dependencias (qué actividad debe terminar antes de que otra pueda empezar). Cada nodo es una actividad y las flechas muestran el orden lógico de ejecución. Es la base sobre la que se calcula todo lo demás — sin la red no puedes calcular ruta crítica ni holguras.

2. Ruta crítica

Es la secuencia de actividades que representa el camino más largo dentro de la red y que determina la duración mínima total del proyecto. Cualquier retraso en una actividad de la ruta crítica retrasa el proyecto completo — por eso se dice que tiene holgura cero.

3. Cálculo hacia adelante y hacia atrás

Para encontrar la ruta crítica se hacen dos pasadas por la red:

Hacia adelante: calcula el inicio y finalización más temprana posible de cada actividad, empezando desde el inicio del proyecto.
Hacia atrás: calcula el inicio y finalización más tardía posible sin afectar la fecha final, empezando desde el final del proyecto.
4. Holgura y retraso
Holgura: la diferencia entre la fecha más temprana y la más tardía en que una actividad puede terminar sin afectar el proyecto. Las actividades con holgura cero forman la ruta crítica.
Las actividades con holgura sí tienen margen — puedes retrasarlas dentro de ese margen sin afectar la fecha final.
5. ¿Qué tan flexible es el sistema?

Explorar la flexibilidad significa identificar cuánta holgura hay disponible en cada actividad no crítica. Un proyecto con mucha holgura en varias rutas es más flexible ante imprevistos; un proyecto con muchas actividades en la ruta crítica (o varias rutas casi críticas) es rígido y de alto riesgo.

6. Técnicas de compresión de cronograma

Cuando el proyecto va atrasado o necesitas terminar antes, hay dos técnicas principales:

Intensificación (crashing): inyectar más recursos (dinero, personal) a las actividades críticas para acortarlas. Aumenta el costo, pero el riesgo se mantiene igual.
Aceleración (fast tracking): ejecutar actividades en paralelo que normalmente irían en secuencia. No aumenta el costo directamente, pero sí aumenta el riesgo y la probabilidad de retrabajo, porque estás empezando algo antes de tener toda la información de la actividad previa.

Regla práctica: intensificación = pagas con dinero; aceleración = pagas con riesgo.

7. Costo del retraso vs. costo de comprar tiempo

Cada día de retraso en la ruta crítica tiene un costo (penalizaciones, pérdida de mercado, etc.). Comparado contra ese costo, se decide si conviene "comprar tiempo" (inyectar capital vía intensificación) — la decisión es puramente de costo-beneficio: si el costo de acelerar es menor al costo de retrasarse, se justifica invertir.

8. Cronograma optimizado (Diagrama de Gantt final)

El resultado de todo este análisis se plasma en un Gantt que muestra:

La ruta crítica resaltada (holgura cero) — son las barras que no se pueden mover.
Las holguras visibles en las demás actividades — para saber dónde hay margen de maniobra.
Hitos de lanzamiento alineados con las fechas clave.
Un plan de contingencia basado en los riesgos identificados (por ejemplo, en tu imagen 4, la actividad "Review" aparece con una alerta de riesgo pendiente aunque tenga las validaciones OK).

Scrum
Es un marco de trabajo (framework), no una metodología rígida.
Ejemplo de aplicación: "Cancha Ya" — proyecto para que jugadores encuentren cancha (ejemplo usado en clase).
Planning Poker: técnica de estimación donde el equipo hace lluvia de ideas individual y luego llega a un acuerdo sobre el esfuerzo de cada tarea.
3 roles:
Product Owner: representa al cliente, define qué se construye.
Scrum Master: guía al equipo y elimina obstáculos.
Equipo de desarrollo: construye el producto.
Se apoya en principios básicos de transparencia, inspección y adaptación.
Kanban

Se centra en la gestión visual del flujo de trabajo:

Visualizar el flujo de trabajo.
Limitar el WIP (Work In Progress / trabajo en curso).
Gestionar el flujo.
Definir políticas explícitas.
Mejorar continuamente (mejora colaborativa/evolutiva).
XP (Extreme Programming)

Se basa fundamentalmente en la comunicación constante entre el equipo y con el cliente, además de prácticas técnicas como programación en pares e integración continua.

TDD (Test-Driven Development)

Ciclo basado en "colores": Rojo (escribir una prueba que falla) → Verde (escribir el código mínimo para que pase) → Azul/Refactor (mejorar el código sin romper la prueba).

Lean

Su principio central es eliminar el desperdicio (todo lo que no agrega valor al cliente).

Crystal

No es una sola metodología, sino una familia de metodologías. El color (Clear, Yellow, Orange, Red, etc.) depende del tamaño del equipo y la criticidad del proyecto.

FDD (Feature-Driven Development)

Desarrollo guiado por funcionalidades (features): se planea, diseña y construye el sistema organizado alrededor de las funcionalidades que debe entregar.

DSDM (Dynamic Systems Development Method)

Es una metodología ágil enfocada en entregar rápido y dentro de plazos fijos, priorizando el tiempo y el presupuesto por encima del alcance (al revés de como se suele planear tradicionalmente).

Principio clave: el tiempo y el costo se fijan desde el inicio; lo que varía es el alcance (las funcionalidades se priorizan y ajustan para caber en esos plazos).
Técnica de priorización — MoSCoW:
Must have (debe tener)
Should have (debería tener)
Could have (podría tener)
Won't have (no tendrá, por ahora)
Se usa mucho en proyectos donde los plazos son inamovibles (fechas de lanzamiento, regulaciones, eventos) y se necesita garantizar que al menos lo esencial ("Must have") se entregue sí o sí.
Combina disciplina de gestión tradicional con la flexibilidad ágil — por eso se considera uno de los frameworks ágiles más "estructurados" del grupo.
