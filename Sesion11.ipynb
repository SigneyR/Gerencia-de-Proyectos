Resumen: Viaje del inversionista – 8.ª edición, planificación y PMBOK 2.4
La idea central es entender que un proyecto de desarrollo de software o infraestructura no se justifica únicamente porque pueda construirse, sino porque genera valor para la organización. Por eso, además de planificar el trabajo técnico, es necesario analizar de dónde provienen los recursos, cuánto cuesta el proyecto, qué beneficios genera y si financieramente es viable.

1. Valor del proyecto: más allá del código
Un proyecto debe enfocarse en generar valor, no simplemente en entregar código o funcionalidades.

El producto final debe resolver una necesidad o problema del negocio.

Es necesario justificar de dónde sale el dinero para financiar el proyecto.

El objetivo no es solamente recortar costos, sino maximizar el valor obtenido con los recursos disponibles.

Un desarrollo puede ser técnicamente exitoso y aun así no generar suficiente valor para justificar la inversión.

Por eso, antes y durante el proyecto se deben relacionar:

Inversión

Costos

Beneficios

Riesgos

Valor generado

Idea clave: hacer el proyecto correctamente no es suficiente; debe ser el proyecto correcto desde el punto de vista del negocio.

2. CAPEX vs. OPEX
Una parte importante de la planificación financiera es diferenciar entre gastos de capital (CAPEX) y gastos operativos (OPEX).

CAPEX — Capital Expenditure

Son inversiones destinadas a adquirir o mejorar activos que tendrán utilidad durante un período prolongado.

Ejemplos: servidores propios, infraestructura, equipos, instalaciones.

Normalmente se realiza como una inversión inicial o de largo plazo.

OPEX — Operational Expenditure

Son gastos necesarios para mantener la operación funcionando.

Ejemplos: servicios cloud, licencias por suscripción, mantenimiento, soporte y servicios recurrentes.

Se generan continuamente durante la operación.

Idea clave: esta diferencia permite entender cómo se financia y cómo se mantiene la infraestructura.

3. Métricas de viabilidad financiera
Para determinar si una inversión tiene sentido financiero se pueden utilizar diferentes indicadores.

VAN — Valor Actual Neto
El VAN calcula cuánto valor genera una inversión considerando el valor del dinero en el tiempo.

VAN > 0: el proyecto genera valor bajo los supuestos utilizados.

VAN = 0: la inversión recupera el valor considerado mediante la tasa de descuento.

VAN < 0: los flujos proyectados no compensan la inversión bajo esos supuestos.

La idea fundamental es que $1 hoy no tiene el mismo valor que $1 en el futuro.

TIR — Tasa Interna de Retorno
La TIR es la tasa de descuento que hace que el VAN sea igual a cero.

Sirve para expresar la rentabilidad implícita del proyecto como un porcentaje.

ROI — Retorno sobre la inversión
El ROI relaciona la ganancia obtenida con la inversión realizada.

Una fórmula simplificada es:

ROI = (Beneficio neto / Inversión) × 100

Permite observar cuánto retorno se obtiene en relación con lo invertido.

Periodo de recuperación — Payback
Indica cuánto tiempo tarda el proyecto en recuperar la inversión inicial mediante los flujos de caja generados.

Idea clave: ninguna métrica debería analizarse completamente aislada; cada una responde a una pregunta financiera diferente.

4. Modelo de flujo de caja
El flujo de caja permite representar cómo entra y sale el dinero durante la vida del proyecto.

Un modelo básico incluye:

Elemento	Qué representa
Ingresos	Dinero que genera el proyecto
Costos	Dinero necesario para desarrollar y operar
Flujo neto	Ingresos − costos
Tasa de descuento	Valor que permite llevar flujos futuros a valor presente

De forma simplificada:

Flujo neto = Ingresos − Costos

Posteriormente, los flujos futuros pueden descontarse para calcular el valor presente y obtener métricas como el VAN.

Ejemplo conceptual:

Invierto $100 millones hoy → durante varios años el proyecto genera ingresos y costos → calculo los flujos netos → descuento los flujos futuros → determino si la inversión genera valor.

Caso de negocio — Business Case
Es el documento que justifica por qué la organización debería realizar el proyecto.

Relaciona aspectos como:

Problema/necesidad → solución → inversión → costos → beneficios → riesgos → valor esperado

El caso de negocio permite que quienes toman decisiones entiendan por qué se solicita el dinero y qué valor se espera obtener a cambio.
