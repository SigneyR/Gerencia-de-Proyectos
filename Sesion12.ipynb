**Resumen sobre el Estándar Global para la Gestión de Proyectos, Monitoreo y Gestión del Cambio **

1.** El Desafío:** Cambio InesperadoOcurre cuando un proyecto avanza a buen ritmo pero surge una necesidad crítica e inmediata a última hora.Para evitar el caos o el descarrilamiento del proyecto, se aplican los principios y directrices de la guía del PMBOK.

2. **Monitoreo y Control **(Cronograma y Finanzas)Funciona como un sistema de navegación ininterrumpido que analiza los datos de desempeño laboral.Permite medir variaciones en tiempo real para calcular desviaciones de tiempo y costo.Asegura una toma de decisiones informada y mantiene una línea del tiempo realista ajustada a los avances de la obra o desarrollo.

3. C**omité de Control de Cambios (CCB)**Es el grupo encargado de evaluar, aprobar, aplazar o rechazar formalmente las solicitudes de cambio.Mantiene el control sobre el alcance y la prioridad de los elementos en el Product Backlog.

4. **Tutorial Práctico:** Proceso de Gestión de SolicitudesIdentificar: Documentar formalmente la necesidad (ej. integración de una pasarela de pago).Evaluar: Analizar minuciosamente el impacto estimado en costo y tiempo (ej. $3,000 USD de impacto).Presentar: Elevar la solicitud formal (Formal Change Request) ante el CCB.Decidir: El comité toma la resolución definitiva (Aprobar, Aplazar o Rechazar).
