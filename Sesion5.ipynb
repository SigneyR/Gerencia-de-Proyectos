Fundamentos: Alcance y Requisitos

Alcance del Producto: describe las características y funciones del resultado final.
Alcance del Proyecto: abarca todo el trabajo realizado para entregar dicho producto.
Requisitos Funcionales: describen lo que el software debe hacer — las capacidades y comportamientos específicos que el sistema debe poseer para satisfacer las necesidades del negocio.
Requisitos No Funcionales: complementan a los funcionales especificando criterios de operación como seguridad, confiabilidad, rendimiento, durabilidad y facilidad de mantenimiento.
Técnicas de Recopilación de Requisitos: se usan métodos como entrevistas, grupos de discusión (focus groups), cuestionarios, análisis de ideas (brainstorming) y análisis de documentos para documentar las necesidades de los interesados.

Tutorial Paso a Paso: Redacción Efectiva

Redactar Historias de Usuario (User Stories) — formato: "Como [rol], quiero [objetivo] para [motivación]". Ejemplo: "Como usuario, quiero recuperar mi contraseña para mantener la seguridad de mi cuenta de forma autónoma." Se recomienda usar formato conversacional, siempre desde la perspectiva del usuario final.
Definir Criterios de Aceptación SMART — condiciones que debe cumplir el entregable para ser aceptado:
Específica
Medible
Alcanzable
Realista
T (plazo definido)
Aplicar la Definición de Terminado (DoD) — checklist final antes del lanzamiento: código completado, pruebas pasadas, documentación actualizada, aprobación del Product Owner.

Documentos Prácticos y Entregables

Matriz de Trazabilidad de Requisitos (RTM): vincula cada requisito desde su origen hasta el entregable final, asegurando que cada elemento aporte valor y permitiendo gestionar cambios de manera asertiva. Columnas típicas: ID único, fuente, prioridad, estado, entregable asociado.
Backlog del Producto Inicial: lista dinámica y priorizada de requisitos, funcionalidades e historias de usuario que representa la visión general y el alcance evolutivo del producto en entornos ágiles (ordenado por prioridad: alta, media, baja).
Alineación con la Entrega de Valor: según PMBOK® 8, el alcance debe estar intrínsecamente ligado a la Estrategia de Desglose de Valor (EDV), para priorizar lo que genera mayor impacto para el cliente.
