Este cronograma documenta el ciclo completo del proyecto, desde el inicio y planeación hasta el cierre, distribuido en 30 actividades organizadas en 10 fases: Inicio y Planeación, Consulta a Expertos, Diseño, Implementación de Servidores, Desarrollo de Software, Pruebas y Testeo, Entrega, Despliegue, Capacitación y Cierre. Cada actividad fue definida, secuenciada y estimada siguiendo el proceso técnico de gestión de cronogramas del PMBOK, especificando su duración, sus relaciones lógicas con otras actividades (Fin-a-Inicio e Inicio-a-Inicio), los ajustes de adelanto o retraso necesarios para optimizar los tiempos sin romper la lógica del proyecto, y el tipo de dependencia (interna o externa, esta última asociada a proveedores y consultores). El resultado es un cronograma de aproximadamente 14.5 semanas, del 21 de septiembre de 2026 al 1 de enero de 2027, representado tanto en una lista detallada de actividades como en un diagrama de Gantt que permite visualizar la duración total, el trabajo en paralelo y los puntos de mayor riesgo del proyecto.
