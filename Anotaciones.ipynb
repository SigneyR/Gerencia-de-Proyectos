**Resumen**

El PMBOK, en su octava edición, representa una evolución en la forma de gestionar los proyectos. A diferencia de versiones anteriores, busca ser más flexible y adaptarse mejor a las características de cada proyecto.

Uno de los conceptos importantes es el tailoring, que podemos entender como “sastrería”. La idea es que no debemos aplicar todos los procesos de la misma manera a todos los proyectos, sino seleccionar y organizar las prácticas que realmente necesitamos según el contexto.

También se habla de trabajar pensando en el proyecto y en la mejora continua. Es decir, ya no se trata solamente de seguir un proceso que ya existe, sino que, cuando todavía no tenemos un proceso definido, debemos crearlo, adaptarlo y mejorarlo durante el desarrollo del proyecto.

En cuanto a la gestión, seguimos teniendo aspectos fundamentales como el alcance, el costo y el cronograma, porque necesitamos controlar qué vamos a hacer, cuánto va a costar y en cuánto tiempo lo vamos a realizar.

Finalmente, está el indicador TCPI, que nos permite conocer la velocidad o nivel de eficiencia que necesitamos mantener de ahora en adelante para alcanzar nuestra meta. Un valor de 1.0 representa una eficiencia necesaria equivalente a la planificada. Si el TCPI es mayor a 1.0, significa que necesitamos ser más eficientes que antes, por lo que funciona como una alarma preventiva para saber que debemos tomar acciones correctivas.

En resumen, el PMBOK 8 busca que la gestión de proyectos sea más adaptable, práctica y orientada a generar valor, utilizando los principios y prácticas que realmente necesita cada proyecto.
