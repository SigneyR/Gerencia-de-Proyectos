Caso de negocio (Business Case)
Es el punto de partida: justifica por qué vale la pena el proyecto antes de invertir recursos.

Problema u oportunidad de negocio Alternativas evaluadas (incluida la de "no hacer nada") Beneficios esperados (cuantificables cuando sea posible: ROI, VAN, TIR, payback) Costos estimados y supuestos Riesgos de alto nivel Recomendación y criterios de éxito

Es el documento que ancla todo lo demás a la generación de valor: si un entregable posterior no se conecta con algo del caso de negocio, es una señal de alerta.

Matriz RAM / RACI
Define quién hace qué en el proyecto, evitando ambigüedad de roles.

Responsable (ejecuta), Aprobador (rinde cuentas), Consultado, Informado Se cruza con la EDT/WBS: cada paquete de trabajo tiene su fila en la matriz Útil especialmente cuando hay múltiples equipos o proveedores 3. Estructura ART (Agile Release Train)

Viene del marco SAFe (Scaled Agile Framework), para escalar agilidad a nivel de programa.

Agrupa varios equipos ágiles (5–12 típicamente) que entregan valor de forma sincronizada Trabaja en Program Increments (PI), ciclos de ~8-12 semanas con iteraciones internas Roles clave: Release Train Engineer (RTE), Product Management, System Architect Eventos: PI Planning, Scrum of Scrums, System Demo, Inspect & Adapt 4. Diagramas de Gantt

Visualización temporal de tareas, dependencias y duraciones.

Útiles en enfoques predictivos/híbridos, menos centrales en agilidad pura Muestran ruta crítica, hitos y solapamientos En contextos ágiles a veces se sustituyen o complementan con roadmaps o release plans 5. Tableros de Valor Ganado (EVM – Earned Value Management)

Mide desempeño combinando alcance, costo y tiempo, no solo "gasto vs. presupuesto". Métricas clave:

PV (Planned Value), EV (Earned Value), AC (Actual Cost) CPI = EV/AC → eficiencia de costo SPI = EV/PV → eficiencia de cronograma EAC/ETC → proyecciones de costo final

Permite detectar desviaciones temprano y tomar acción correctiva basada en datos, no en percepción.

Registro de riesgos
Documento vivo que se actualiza durante todo el proyecto.

Identificación, categoría, probabilidad, impacto, prioridad (matriz P×I) Estrategia de respuesta: evitar, mitigar, transferir, aceptar Dueño del riesgo y plan de contingencia Estado (abierto, monitoreado, cerrado, materializado) 7. Artefactos de previsibilidad (Predictability)

Especialmente relevantes en SAFe/ART: miden qué tan confiable es el equipo/tren para cumplir lo que se compromete.

Program Predictability Measure: % de objetivos de negocio del PI realmente logrados vs. planeados Burn-up/burn-down a nivel de PI Velocidad histórica de los equipos (para calibrar futuros compromisos) Esto retroalimenta directamente el caso de negocio: si la previsibilidad es baja, el valor prometido no se está materializando en los tiempos esperados Cómo se conecta todo con "creación sistemática de valor"

La lógica de fondo es un ciclo continuo:

Caso de negocio (define el valor esperado) → RAM/ART (organiza quién lo entrega) → Gantt/Backlog (planifica cómo y cuándo) → EVM/Previsibilidad (mide si el valor se está entregando según lo planeado) → Registro de riesgos (protege esa entrega de valor) → esto retroalimenta ajustes al caso de negocio o al plan.

En otras palabras: no son entregables aislados de "gerencia de proyectos" por separado, sino un sistema de control de valor: el caso de negocio pone la meta, y el resto de artefactos existen para verificar, en cada punto del proyecto, si esa meta de valor sigue siendo alcanzable y se está cumpliendo.
